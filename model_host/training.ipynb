```python
import sklearn
import numpy as np
import pandas as pd

# chunk data processing
def construct_array(input, chunk_size):
    arr = np.empty((0, 11))
    tmp = []
    with open(input, "r") as file:
        for line in file: # assume clean input (11 letters)
            line = line.rstrip() # newline at end
            if len(tmp) == chunk_size:
                np_tmp = np.array(tmp)
                arr = np.vstack((arr, np_tmp))
                tmp = []

            vals = line.split(" ") # len==11
            for i in range(10): 
                vals[i] = ord(vals[i]) - ord("A") + 1 # A as 1, B as 2 ...
            vals[10] = 0 if vals[10] == "I" else 1 # E as 1, I as 0
            tmp.append(vals)
        
        # if leftover 
        if len(tmp) > 0:
            np_tmp = np.array(tmp)
            arr = np.vstack((arr, np_tmp))
    return arr

np_data = construct_array("gen_data_10000.txt", 30)
# np_data

col_labels = ["Q1", "Q2", "Q3", "Q4", "Q5", "Q6", "Q7", "Q8", "Q9", "Q10", "Res"]
df = pd.DataFrame(np_data, columns=col_labels)
df

# train test split etc -> train
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

X = df.iloc[:, :10].values
y = df.iloc[:, 10].values
# print(np.count_nonzero(y_train == 1))
# y_train
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25)
# X_train.shape
model = LogisticRegression(random_state=1).fit(X_train, y_train)

# print(model.predict(np.array([1, 1, 1, 2, 2, 2, 4, 3, 4, 4]).reshape(1, -1)))

model.score(X_test, y_test)
# scores perfectly, likely due to the fact that the data is generated -> a line can 100% be drawn
# can increase difficulty of model by adding multiclass
```
